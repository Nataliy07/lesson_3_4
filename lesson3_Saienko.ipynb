Стандартне відхилення (std)

```python
w = df["Lines_of_Code"]
print(w.mean())
print(w.std())
```

Правило пальця: приблизно дві третини значень лежать у межах «середнє ± std». Перевіримо на даних.

```python
low, high = w.mean() - w.std(), w.mean() + w.std()
print(round(low), round(high))
print(((w >= low) & (w <= high)).mean())   
```

Середнє 639.46, std 253.07: від 386 до 893 кількість рядків коду. Орієнтовно 68,3 %, тобто близько двох третин.

### 3.9 Кореляція: чи рухаються величини разом

```python
print(df["AI_Usage_Hours"].corr(df["Lines_of_Code"]))
print(df["Sleep_Hours"].corr(df["Stress_Level"]))
```

0.86 кореляція сильна позитивна. Значення дуже близьке до +1, що свідчить про чітку узгодженість між двома показниками, тобто чим більше коду - тим більше годин.
-0.75 сильний зворотний (негативний) лінійний зв'язок. Чим менше сну, тим вищий стрес та навпаки.
